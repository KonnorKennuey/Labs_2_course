## Задание 1. GitHub API

Получим список репозиториев пользователя и сохраним ответ в JSON-файл.

```python
import json
import requests
```

```python
github_username = 'KonnorKennuey'
github_url = f'https://api.github.com/users/{github_username}/repos'
github_response = requests.get(github_url, params={'per_page': 100})

if github_response.status_code == 200:
    github_repositories = github_response.json()
    print('Репозитории:')
    for repository in github_repositories:
        print(repository['name'])

    with open('github_repositories.json', 'w', encoding='utf-8') as file:
        json.dump(github_repositories, file, ensure_ascii=False, indent=4)

    print('Ответ сохранён в github_repositories.json')
else:
    print('Ошибка:', github_response.status_code, github_response.text)
```
